### 📋 Zadanie 1: Wektorowy podział rabatowy z np.where()

Masz serię kwot transakcji: `kwoty = pd.Series([120, 450, 1200, 3100, 85, 2400])`.
Napisz kod wykorzystujący `np.where()`, który dla kwot powyżej 1000 zł nalicza rabat 15%, a dla pozostałych 0 zł.


```python
# TUTAJ WPISZ SWÓJ KOD:
import pandas as pd
import numpy as np

kwoty = pd.Series([120, 450, 1200, 3100, 85, 2400])
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
rabaty = np.where(kwoty > 1000, kwoty * 0.15, 0.0)
rabaty = pd.Series(rabaty)
print('Kwota', kwoty)
print('Rabat', rabaty)

# data frame:
df = pd.DataFrame({"Kwoty": kwoty, "Rabaty": rabaty})
print(df)
```

### 📋 Zadanie 2: Segmentacja klientów z np.select()

Dla ramki `df_klient = pd.DataFrame({'Wiek': [20, 45, 72, 33, 17]})` przypisz kolumnę `'Kategoria'`: `'Junior'` dla wieku < 18, `'Dorosły'` dla 18-65 oraz `'Senior'` dla > 65.


```python
# TUTAJ WPISZ SWÓJ KOD:
df_klient = pd.DataFrame({'Wiek': [20, 45, 72, 33, 17]})
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
warunki = [df_klient['Wiek'] < 18, df_klient['Wiek'] <= 65]
etykiety = ['Junior', 'Dorosły']
df_klient['Kategoria'] = np.select(warunki, etykiety, default='Senior')
print(df_klient)
```
