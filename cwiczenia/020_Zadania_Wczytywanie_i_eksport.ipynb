### 📋 Zadanie 1: Parsowanie danych z polskim przecinkiem

Mając tekst w formacie CSV z separatorem `;` i przecinkiem jako separatorem dziesiętnym, wczytaj go do DataFrame upewniając się, że kolumna z ceną jest typu `float64`.


```python
# TUTAJ WPISZ SWÓJ KOD:
import pandas as pd
import io
```

```python

surowe_dane = '''produkt;cena_zl
Kawa;14,99
Herbata;8,50
Cukier;4,20'''
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
df_prod = pd.read_csv(io.StringIO(surowe_dane), sep=';', decimal=',')
print(df_prod)
print('Typ ceny:', df_prod['cena_zl'].dtype)
```

### 📋 Zadanie 2: Rozwijanie zagnieżdżonej listy zamówień z .explode()

Mając tabelę z kolumną zawierającą listy artykułów, przekształć ją metodą `.explode()`, aby każdy artykuł znalazł się w osobnym wierszu.


```python
# TUTAJ WPISZ SWÓJ KOD:
df_zam = pd.DataFrame({
    'ID': [1, 2],
    'Klient': ['Jan', 'Anna'],
    'Koszyk': [['Mleko', 'Chleb'], ['Woda']]
})
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
df_rozbite = df_zam.explode('Koszyk').reset_index(drop=True)
print(df_rozbite)
```
