### 📋 Zadanie 1: Statystyki, ekstrema i brakujące dane

Dla serii ocen: `s = pd.Series([4.5, 3.0, 5.0, 2.0, np.nan, 5.0, 4.0])`:
1. Oblicz liczbę braków za pomocą `.isnull().sum()`.
2. Wypełnij brakujące wartości średnią ocen.
3. Wyznacz 2 najwyższe oceny za pomocą metody `.nlargest(2)`.


```python
# TUTAJ WPISZ SWÓJ KOD:
import pandas as pd
import numpy as np

s = pd.Series([4.5, 3.0, 5.0, 2.0, np.nan, 5.0, 4.0])
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
print('Liczba braków:', s.isnull().sum())
s_uzupelnione = s.fillna(s.mean())
print('Top 2 oceny:\n', s_uzupelnione.nlargest(2))
```

### 📋 Zadanie 2: Slicing i resetowanie indeksu z drop=True
