### 📋 Zadanie 1: Selektywne czyszczenie braków z parametrem subset

W tabeli zamówień usuń TYLKO te wiersze, w których brakuje kwoty transakcji (`subset=['kwota']`). Rekordy z brakującym emailem powinny pozostać w bazie.


```python
# TUTAJ WPISZ SWÓJ KOD:
import pandas as pd
import numpy as np

df_zam = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'klient': ['Jan', 'Anna', 'Piotr', 'Maria'],
    'kwota': [150.0, np.nan, 220.0, np.nan],
    'email': ['jan@a.pl', None, 'piotr@c.pl', None]
})
# Twój kod poniżej:
df_zam
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
df_czyste = df_zam.dropna(subset=['kwota'], ignore_index=True)
print(df_czyste)
```

### 📋 Zadanie 2: Deduplikacja z zachowaniem najnowszego rekordu (keep='last')

Posortuj logi według kolumny `'data'` i usuń duplikaty użytkownika (`subset=['user']`), zachowując wyłącznie jego ostatnie logowanie (`keep='last'`).


```python
# TUTAJ WPISZ SWÓJ KOD:
df_logi = pd.DataFrame({
    'user': ['jan', 'anna', 'jan', 'piotr'],
    'data': ['2024-01-01', '2024-01-02', '2024-01-05', '2024-01-03'],
    'ip': ['1.1.1.1', '2.2.2.2', '1.1.1.9', '3.3.3.3']
})
# Twój kod poniżej:
df_logi
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
df_posortowane = df_logi.sort_values(by='data')
df_unikalne = df_posortowane.drop_duplicates(subset=['user'], keep='last', ignore_index=True)
print(df_unikalne)
```

### 📋 Zadanie 3: Sortowanie wielokolumnowe z różnymi kierunkami

Posortuj poniższą tabelę tak, aby `'Dzial'` był posortowany rosnąco (A-Z), a `'Pensja'` malejąco (od najwyższej).


```python
# TUTAJ WPISZ SWÓJ KOD:
df_pracownicy = pd.DataFrame({
    'Dzial': ['IT', 'Marketing', 'IT', 'Marketing', 'Finanse'],
    'Pracownik': ['Jan', 'Ewa', 'Anna', 'Piotr', 'Zofia'],
    'Pensja': [8000, 7500, 14000, 11000, 15000]
})
# Twój kod poniżej:
df_pracownicy
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
df_posortowani = df_pracownicy.sort_values(by=['Dzial', 'Pensja'], ascending=[True, False], ignore_index=True)
print(df_posortowani)
```

### 📋 Zadanie 4: Bezpieczna iteracja z .itertuples()

Napisz pętlę wykorzystującą `df_pracownicy.itertuples()`, która wyświetli tekst: `'Pracownik: {Pracownik} zarabia {Pensja} zł w dziale {Dzial}'`.


```python
# TUTAJ WPISZ SWÓJ KOD:
# Twój kod poniżej:
for pracownik in df_pracownicy.itertuples():
    print(
        f"Pracownik: {pracownik.Pracownik} zarabia {pracownik.Pensja} zł w dziale {pracownik.Dzial}"
    )
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
for r in df_pracownicy.itertuples():
    print(f'Pracownik: {r.Pracownik} zarabia {r.Pensja} zł w dziale {r.Dzial}')
```
