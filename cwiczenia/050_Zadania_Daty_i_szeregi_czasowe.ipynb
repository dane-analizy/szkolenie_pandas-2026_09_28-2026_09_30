### 📋 Zadanie 1: Parsowanie dat z błędami (errors='coerce')

Skonwertuj listę wpisów z formularza do `pd.to_datetime` z `errors='coerce'` i wyświetl wyłącznie rekordy, które okazały się błędnymi datami (`NaT`).


```python
# TUTAJ WPISZ SWÓJ KOD:
import pandas as pd
import numpy as np

wpisy = ['2024-01-15', '12.02.2024', 'BŁĄD', '2024-05-30', 'nie dotyczy']
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
daty = pd.to_datetime(wpisy, errors="coerce", format="mixed", dayfirst=True)
uszkodzone = [w for w, d in zip(wpisy, daty) if pd.isna(d)]
print('Uszkodzone wpisy:', uszkodzone)
```

```python
pd.Series(wpisy)[pd.isna(daty)]
```

### 📋 Zadanie 2: Resampling dzienny sumy transakcji

Mając transakcje co 30 minut, zagreguj je do podsumowań dziennych metodą `.resample('D').sum()`.


```python
# TUTAJ WPISZ SWÓJ KOD:
import numpy as np

idx = pd.date_range('2024-01-01', periods=48*5, freq='30min')
df_trans = pd.DataFrame({"Kwota": np.random.uniform(50, 200, size=48 * 5)}, index=idx)
# Twój kod poniżej:
```

```python
df_trans
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
dzienne = df_trans.resample('D').sum()
print(dzienne)
```
