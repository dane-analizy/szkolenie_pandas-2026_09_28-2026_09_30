### 📋 Zadanie 1: Grupowanie z NamedAgg i as_index=False

Dla tabeli sprzedaży oblicz per `'Region'`: łączny obrót (`'Suma'`) oraz średnią wartość zamówienia (`'Srednia'`), zachowując czyste kolumny bez MultiIndex.


```python
# TUTAJ WPISZ SWÓJ KOD:
import pandas as pd

df_sprzedaz = pd.DataFrame({
    'Region': ['Północ', 'Północ', 'Południe', 'Wschód', 'Południe'],
    'Wartosc': [120, 450, 300, 80, 500]
})
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
raport = df_sprzedaz.groupby('Region', as_index=False).agg(
    Suma=('Wartosc', 'sum'),
    Srednia=('Wartosc', 'mean')
)
print(raport)
```

### 📋 Zadanie 2: Złączenie relacyjne z kontrolą relacji (validate='1:m')

Połącz tabelę klientów z tabelą zamówień metodą `how='left'`, upewniając się za pomocą `validate='1:m'`, że w tabeli klientów klucz `'id'` jest unikalny.


```python
# TUTAJ WPISZ SWÓJ KOD:
klienci = pd.DataFrame({'id': [1, 2], 'nazwisko': ['Kowalski', 'Nowak']})
zamowienia = pd.DataFrame({'zam_id': [101, 102, 103], 'id': [1, 1, 2], 'kwota': [100, 200, 300]})
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
wynik = pd.merge(klienci, zamowienia, on='id', how='left', validate='1:m', indicator=True)
print(wynik)
```
