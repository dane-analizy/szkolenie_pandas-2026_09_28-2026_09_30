### 📋 Zadanie 1: Wykres słupkowy w Plotly Express

Zbuduj interaktywny wykres słupkowy (`px.bar`) przedstawiający przychody 4 działów firmy i dodaj etykiety wartości na słupkach (`text_auto=True`).


```python
# TUTAJ WPISZ SWÓJ KOD:
import pandas as pd
import plotly.express as px

df_dzialy = pd.DataFrame({
    'Dzial': ['IT', 'HR', 'Finanse', 'Sprzedaż'],
    'Przychod': [120000, 45000, 89000, 150000]
})
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
fig = px.bar(df_dzialy, x='Dzial', y='Przychod', text_auto=True, title='Przychody Działów')
# fig.show()
print('Wykres wygenerowany pomyślnie!')
```
