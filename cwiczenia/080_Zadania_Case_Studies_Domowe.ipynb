```python
import os
import pandas as pd
import numpy as np

# Wczytanie danych olimpijskich do zadań 1-5:
if os.path.exists("../olimpics.zip"):
    df_raw = pd.read_csv("../olimpics.zip")
elif os.path.exists("olympics_dataset.csv"):
    df_raw = pd.read_csv("olympics_dataset.csv")
else:
    df_raw = pd.read_csv("olympics.csv")

df_olimp = df_raw.rename(columns={"Name": "Athlete", "Team": "Country"}).copy()
if "Medal" in df_olimp.columns and "No medal" in df_olimp["Medal"].values:
    df_olimp = df_olimp[df_olimp["Medal"] != "No medal"]

print("Załadowano bazę olimpijską, rekordów medalowych:", len(df_olimp))
```

### 🏅 Zadanie Olimpijskie 1: Dynamika równouprawnienia płci na podium

Dla każdej edycji igrzysk w ramce `df_olimp` oblicz procentowy udział kobiet wśród wszystkich medalistów danej edycji.
Wskaż 3 edycje o najniższym oraz 3 edycje o najwyższym udziale kobiet.


```python
# TUTAJ WPISZ SWÓJ KOD:

```

#### 💡 Rozwiązanie wzorcowe:


```python
# ROZWIĄZANIE WZORCOWE:
plec_lata = df_olimp.groupby(["Year", "Sex"]).size().unstack(fill_value=0)
plec_lata["Udzial_Kobiet"] = (plec_lata["F"] / (plec_lata["F"] + plec_lata["M"]) * 100).round(1)

print("3 edycje o najniższym udziale kobiet:")
print(plec_lata["Udzial_Kobiet"].sort_values().head(3))

print("\n3 edycje o najwyższym udziale kobiet:")
print(plec_lata["Udzial_Kobiet"].sort_values(ascending=False).head(3))
```

### 🏅 Zadanie Olimpijskie 2: Ważony ranking medalowy w pływaniu (Swimming)

Przyjmując ranking punktowy: Złoto = 6 pkt, Srebro = 2 pkt, Brąz = 1 pkt, wyznacz 10 najbardziej utytułowanych sportowców w dyscyplinie *Swimming*.
Tabela wynikowa powinna zawierać kolumny: `Gold`, `Silver`, `Bronze`, `Punkty` oraz być posortowana malejąco po punktach.


```python
# TUTAJ WPISZ SWÓJ KOD:

```

#### 💡 Rozwiązanie wzorcowe:


```python
# ROZWIĄZANIE WZORCOWE:
plywanie = df_olimp[df_olimp["Sport"] == "Swimming"].pivot_table(
    index=["Athlete", "Country"],
    columns="Medal",
    values="Event",
    aggfunc="count",
    fill_value=0
)[["Gold", "Silver", "Bronze"]]

plywanie["Punkty"] = 6 * plywanie["Gold"] + 2 * plywanie["Silver"] + plywanie["Bronze"]
print(plywanie.sort_values(by="Punkty", ascending=False).head(20))
```

### 🏅 Zadanie Olimpijskie 4: Paradoks Montrealu 1976 – Deduplikacja konkurencji

W surowej bazie dla Igrzysk w Montrealu (1976) jest aż 73 polskich medalistów, jednak oficjalnie Polska zdobyła 26 medali (7 złotych, 6 srebrnych, 13 brązowych).
Wykorzystaj metodę `drop_duplicates` z odpowiednim parametrem `subset`, aby odtworzyć oficjalny bilans Polski w Montrealu.


```python
# TUTAJ WPISZ SWÓJ KOD:

```

#### 💡 Rozwiązanie wzorcowe:


```python
# ROZWIĄZANIE WZORCOWE:
df_oficjalne = df_olimp.drop_duplicates(subset=["Year", "Country", "Sport", "Event", "Medal"])
polska_1976 = df_oficjalne.query("Country in ['POL', 'Poland'] and Year == 1976")
bilans = polska_1976["Medal"].value_counts()[["Gold", "Silver", "Bronze"]]
print(bilans)
print("Razem medali:", bilans.sum())
```

### 🛒 Projekt Domowy 6: Analiza koszykowa (Współwystępowanie par produktów)

Dla tabeli zamówień `df_koszyk` znajdź pary produktów najczęściej kupowane razem w ramach tego samego `zamowienie_id` (zastosuj self-merge).


```python
# TUTAJ WPISZ SWÓJ KOD:
import pandas as pd

df_koszyk = pd.DataFrame({
    "zamowienie_id": [1, 1, 2, 2, 3, 3, 4, 4, 4],
    "produkt": ["Chleb", "Masło", "Chleb", "Ser", "Chleb", "Masło", "Ser", "Chleb", "Wędlina"]
})
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe:


```python
# ROZWIĄZANIE WZORCOWE:
pary = pd.merge(df_koszyk, df_koszyk, on="zamowienie_id")
pary_unikalne = pary[pary["produkt_x"] < pary["produkt_y"]]
ranking = pary_unikalne.groupby(["produkt_x", "produkt_y"]).size().reset_index(name="Razem")
print(ranking.sort_values(by="Razem", ascending=False))
```

### ⚡ Projekt Domowy 7: Wykrywanie bloków awarii w telemetrii przemysłowej

Mając wektor flag anomalii logicznych, ponumeruj kolejne odrębne incydenty awarii za pomocą metod `.diff()` (lub `.shift()`) oraz `.cumsum()`.


```python
# TUTAJ WPISZ SWÓJ KOD:
import pandas as pd

anomalie = pd.Series([False, True, True, False, True, True, False, True])
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe:


```python
# ROZWIĄZANIE WZORCOWE:
nowe_incydenty = (anomalie == True) & (anomalie.shift(1, fill_value=False) == False)
nr_incydentu = nowe_incydenty.cumsum() * anomalie
print(pd.DataFrame({"Czy_Anomalia": anomalie, "ID_Incydentu": nr_incydentu}))
```
