### 📋 Zadanie 1: Optymalizacja pamięci RAM typem category

Przekonwertuj kolumnę tekstową z powtarzalnymi statusami na typ `'category'` i porównaj zużycie pamięci przed i po konwersji za pomocą `.memory_usage(deep=True)`.


```python
# TUTAJ WPISZ SWÓJ KOD:
import pandas as pd
import numpy as np

s_statusy = pd.Series(np.random.choice(['NOWY', 'W_TRAKCIE', 'ZAMKNIETY'], size=1_000_000))
# Twój kod poniżej:
```

#### 💡 Rozwiązanie wzorcowe (kliknij komórkę poniżej aby podejrzeć):


```python
# ROZWIĄZANIE WZORCOWE:
ram_przed = s_statusy.memory_usage(deep=True) / 1024
s_kat = s_statusy.astype('category')
ram_po = s_kat.memory_usage(deep=True) / 1024

print(f'Przed: {ram_przed:.1f} KB | Po: {ram_po:.1f} KB | Oszczędność: {(1 - ram_po/ram_przed)*100:.1f}%')
```
