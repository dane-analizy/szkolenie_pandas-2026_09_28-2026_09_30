# 030. Praca ze Strukturą Series

Jednowymiarowa tablica etykietowana `pd.Series`: slicing etykietowy vs pozycyjny, zarządzanie indeksem (`reset_index(drop=True)`), metody statystyczne (`.sum()`, `.mean()`, `.nlargest()`), `.describe()`, unikalność (`.unique()`, `.nunique()`), braki (`.isnull().sum()`, `.fillna()`) oraz szybkie mapowanie słownikowe.


## 1. Tworzenie i badanie struktury Series


```python
import pandas as pd
import numpy as np
```

```python
ceny = pd.Series(
    [12.50, 45.00, 8.99, 120.00, 24.50, 65.00, 45.00], 
    index=['Kawa', 'Herbata', 'Woda', 'Wino', 'Sok', 'Miód', 'Ciastko'],
    name='Cena_PLN'
)

print("Liczba elementów (.size):", ceny.size)
print("Typ danych (.dtype):", ceny.dtype)
print("Indeks etykiet (.index):", ceny.index)
print("Unikalne wartości (.unique()):", ceny.unique())
print("Liczba unikalnych cen (.nunique()):", ceny.nunique())
print("\nStatystyki opisowe (.describe()):\n", ceny.describe())
```

## 2. Mistrzostwo Slicingu (`loc` vs `iloc`)


```python
# loc: obustronnie domknięty (zawiera 'Herbata' oraz 'Wino')
print("Wycinek loc ['Herbata':'Wino']:")
print(ceny.loc['Herbata':'Wino'])
```

```python
# iloc: półotwarty (pozycje 1 i 2, bez pozycji 3!)
print("\nWycinek iloc [1:3]:")
print(ceny.iloc[1:3])
```

```python

# Indeksy ujemne i krok:
print("\nTrzy ostatnie produkty:", ceny.iloc[-3:])
print("Co drugi produkt od końca:", ceny.iloc[::-2])
```

## 3. Metody agregujące, statystyczne i ekstrema (`.nlargest()`)


```python
print("Suma:", ceny.sum())
print("Średnia arytmetyczna:", round(ceny.mean(), 2))
print("Mediana:", ceny.median())
print("Odchylenie standardowe (.std()):", round(ceny.std(), 2))
```

```python
# Ekstrema - dużo szybsze od pełnego sortowania:
print("\nTop 3 najdroższe produkty (.nlargest(3)):")
print(ceny.nlargest(3))

print("\n2 najtańsze produkty (.nsmallest(2)):")
print(ceny.nsmallest(2))
```

## 4. Obsługa braków: `.isnull().sum()`, `.dropna()` i `.fillna()`


```python
s_braki = pd.Series([10.0, np.nan, 25.0, None, 40.0, np.nan], index=list('abcdef'))

print("Liczba braków (.isnull().sum()):", s_braki.isnull().sum())
print("Liczba braków (.isna().sum()):", s_braki.isna().sum())
```

```python
# Usuwanie braków
print("\nCzyste (dropna):\n", s_braki.dropna())
```

```python
# Wypełnianie braków średnią oraz propagacja (ffill):
print("\nWypełnione średnią:\n", s_braki.fillna(s_braki.mean()))
print("\nWypełnione w przód (ffill):\n", s_braki.ffill())
```

## 5. Sortowanie, rozkłady częstości i mapowanie


```python
# Sortowanie po wartościach i indeksie
print("Ceny rosnąco:\n", ceny.sort_values(ascending=True))
```

```python

# Badanie częstości z value_counts (z uwzględnieniem procentów)
statusy = pd.Series(['Aktywny', 'VIP', 'Aktywny', 'Nowy', None, 'VIP', 'Aktywny'])
print("\nRozkład procentowy statusów (dropna=False):")
print(statusy.value_counts(normalize=True, dropna=False).round(3) * 100)
```

```python

# Szybka podmiana słownikowa (.map)
kody = pd.Series(['PL', 'DE', 'FR', 'PL'])
print("\nPrzetłumaczone kody (.map):\n", kody.map({'PL': 'Polska', 'DE': 'Niemcy', 'FR': 'Francja'}))
```
