# 070. Wizualizacja Danych w Pandas i Plotly Express

Od szybkiej eksploracji metodą `df.plot()` po interaktywne dashboardy biznesowe w bibliotece **Plotly Express** z eksportem do formatu HTML.


## 1. Szybka eksploracja metodą `df.plot()`


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

df_wyniki = pd.DataFrame({
    'Dzial': ['IT', 'Marketing', 'Finanse', 'Sprzedaż'],
    'Przychod_kPLN': [450, 280, 510, 620]
}).set_index('Dzial')

# Szybki wykres słupkowy
ax = df_wyniki.plot(kind='bar', figsize=(7, 4), color='teal', legend=False)
plt.title("Przychód działów")
plt.ylabel("tys. PLN")
plt.tight_layout()
plt.show()
```

## 2. Interaktywne dashboardy biznesowe w Plotly Express


```python
import plotly.express as px

np.random.seed(42)
dane_klientow = pd.DataFrame({
    'Wiek': np.random.randint(18, 70, 100),
    'Wydatki_Roczne': np.random.uniform(1000, 15000, 100),
    'Liczba_Transakcji': np.random.randint(2, 40, 100),
    'Segment': np.random.choice(['VIP', 'Lojalny', 'Standard'], 100)
})

# Wykres rozrzutu z interaktywnymi dymkami (hover)
fig = px.scatter(
    dane_klientow,
    x='Wiek',
    y='Wydatki_Roczne',
    size='Liczba_Transakcji',
    color='Segment',
    title="Segmentacja Bazy Klientów (Plotly Express)",
    labels={'Wydatki_Roczne': 'Wydatki Roczne (PLN)', 'Wiek': 'Wiek klienta'},
    template='plotly_white'
)
# fig.show()
# fig.write_html('dashboard_klientow.html')
print("Obiekt wykresu Plotly wygenerowany pomyślnie!")
```
