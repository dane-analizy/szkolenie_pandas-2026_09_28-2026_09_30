```python
lista = [1,2,3,4,2,3,4,5]

# czy liczba jest parzysta?
for element in lista:
    if element >= 3:
        print(f"{element} jest >=3 ")
    else:
        print(f"{element} jest < 3")
```

```python
lista
```

```python
np_lista = np.array(lista)
np.where(np_lista >= 3, ">= 3", "< 3")
```

```python
np_lista
```

```python
np.where(np_lista % 2, "parzyste", "nieparzyste")
```

```python
zarobki = np.array(  [3500, 7250, 12500, 4800, 9100 ]  )
```

```python
np.where(zarobki > 8000, "drugi próg", "pierwszy próg")
```

```python
warunki = [
    zarobki < 5000,
    (zarobki >= 5000) & (zarobki < 10000),
    zarobki >= 10000
]
etykiety = ['junior', 'mid', 'senior']
```

```python
warunki
```

```python
np.select( warunki, etykiety, default="nieznany" )
```

```python
zarobki = np.random.normal(10000, 5000, size=1_000_000)
warunki = [zarobki < 5000, (zarobki >= 5000) & (zarobki < 10000), zarobki >= 10000]
etykiety = ["junior", "mid", "senior"]
```

```python
np.select(warunki, etykiety, default="nieznany")
```

```python
zarobki = np.array([3500, 7250, 12500, 4800, 9100])
warunki = [
    zarobki > 3000,
    zarobki < 5000,
    (zarobki >= 5000) & (zarobki < 10000),
    zarobki >= 10000,
]
etykiety = ["junior","junior2", "mid", "senior"]
np.select(warunki, etykiety, default="nieznany")
```
