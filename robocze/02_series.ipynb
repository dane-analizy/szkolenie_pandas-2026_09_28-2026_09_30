```python
# mapowanie
slownik = {
 "PL": "Polska",
 "DE": "Niemcy",
 "FR": "Francja",
 "GB": "Wielka Brytania",
 "IT": "Włochy",
 "ES": "Hiszpania",
 "NL": "Holandia",
 "CH": "Szwajcaria",
 "SE": "Szwecja",
 "NO": "Norwegia",
}

kraje = pd.Series(
    [
        "PL",
        "DE",
        "FR",
        "GB",
        "IT",
        "ES",
        "NL",
        "CH",
        "SE",
        "NO",
        "FR",
        "GB",
        "IT",
        "CH",
        "SE",
        "NO",
        "FR",
        "GB",
        "ES",
        "NL",
        "CH",
        "SE",
        "USA"
    ]
)

kraje
```

```python
# mapowanie skrótów na nazwy
kraje.map(slownik)
```

```python
# mapowanie funkcją
def mapowanie(s):
    if s == "USA":
        return "Ameryka panie!"
    return 1 if s[0] < "K" else 0
```

```python
mapowanie("GB"), mapowanie("PL"), mapowanie("USA")
```

```python
kraje.map(mapowanie)
```

```python
df = pd.DataFrame({"iso_kraju": kraje})
df
```

```python
df['nazwa_kraju'] = df['iso_kraju'].map(slownik)
df
```

```python
slownik_df = pd.DataFrame(slownik.items())
slownik_df.columns = ['iso', 'nazwa']
slownik_df
```

```python
pd.merge(df, slownik_df, how="left", left_on="iso_kraju", right_on="iso")
```

```python
dfa = pd.DataFrame({"masa_wzrost": pd.Series([ [1,2], [3,4], [5,6]])})
dfa
```

```python
def licz_bmi(wektor):
    return wektor[0] * wektor[1]
```

```python
dfa['bmi'] = dfa["masa_wzrost"].map(licz_bmi)
dfa
```

```python
dfa.to_csv("plik.csv")
```

```python
dfa.to_excel("plik.xlsx")
```

```python
dfa['masa'] = dfa['masa_wzrost'].map(lambda w: w[0])
dfa["wzrost"] = dfa["masa_wzrost"].map(lambda w: w[1])
dfa
```

```python
dfa["nowe"] = dfa.apply(lambda row: [row["masa"], row["wzrost"]], axis=1)
dfa
```
