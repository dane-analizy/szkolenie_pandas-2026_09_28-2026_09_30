```python
import pandas as pd
import numpy as np
```

```python
# seria z samymi wartościami  -zwróć uwagę na indeks i nazwe
ceny = pd.Series([12.50, 45.00, 8.99, 120.00, 24.50, 65.00, 45.00])
ceny
```

```python
# seria z  wartościami i indeksem  -zwróć uwagę na indeks i nazwe
ceny = pd.Series(
    [12.50, 45.00, 8.99, 120.00, 24.50, 65.00, 45.00],
    index=["Kawa", "Herbata", "Woda", "Wino", "Sok", "Miód", "Ciastko"],
)
ceny
```

```python
# seria z wartościami, indeksem i nazwą
ceny = pd.Series(
    [12.50, 45.00, 8.99, 120.00, 24.50, 65.00, 45.00],
    index=["Kawa", "Herbata", "Woda", "Wino", "Sok", "Miód", "Ciastko"],
    name="Cena_PLN",
)
ceny
```
