```python
import pandas as pd
```

```python
dane = pd.read_csv("spoleczenstwo.csv", sep="|")
dane
```

```python
dane["imie"].value_counts()
```

```python
dane["nazwisko"].value_counts()
```

```python
nazwiska = pd.DataFrame(dane["nazwisko"].value_counts()).reset_index(drop=False)
nazwiska
```

```python
imiona = pd.DataFrame(dane["imie"].value_counts()).reset_index(drop=False)
```

```python
# to nadpisze plik!
imiona.to_excel("spoleczensto.xlsx", index=False, sheet_name="imiona")
nazwiska.to_excel("spoleczensto.xlsx", index=False, sheet_name="nazwiska")
```

```python
with pd.ExcelWriter("spoleczensto_ok.xlsx", engine="openpyxl") as writer:
        dane.to_excel(writer,  index=False, sheet_name="dane_surowe")
        imiona.to_excel(writer, index=False, sheet_name="imiona")
        nazwiska.to_excel(writer, index=False, sheet_name="nazwiska")
```

```python
dane.to_excel("spoleczensto_full.xlsx", index=False)
```
