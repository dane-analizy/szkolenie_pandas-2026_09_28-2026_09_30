```python
liczba_wierszy = towary.shape[0]
kolumny_niepuste = []

for col in towary.columns:
    print(f"Kolumna: {col}")
    licz_puste = towary[col].isnull().sum()
    print(f"Pustych: {licz_puste}")
    if licz_puste == liczba_wierszy:
        print("Wszystkie wiersze puste!")
    else:
        kolumny_niepuste.append(col)
    print()

print(f"Niepuste kolumny: {kolumny_niepuste}")
```
