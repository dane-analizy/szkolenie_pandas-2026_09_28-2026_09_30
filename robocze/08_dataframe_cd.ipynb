```python
import pandas as pd
import json
from pathlib import Path
```

```python
dane_lista = []
for plik in Path("../dane_przykladowe/pogoda").glob("*.json"):
    with open(plik, "r", encoding="utf-8") as f:
        dane_tmp = json.load(f)
        dane_lista.append(dane_tmp)
```

```python
df = pd.DataFrame(dane_lista)
df
```

```python
df = pd.concat(
    [
        df,
        pd.json_normalize(df["main"]),
        pd.json_normalize(df["wind"])
    ],
    axis=1 # obok
)
df
```

```python
df["weather"].explode()
```

```python
df["weather_exploded"] = df["weather"].explode()
```

```python
df = pd.concat(
    [df, pd.json_normalize(df["weather_exploded"])],
    axis=1,  # obok
)
df
```

```python
# usuwanie kolumn
df = df.drop(columns=["main", "wind", "weather_exploded", "weather"])
df
```
