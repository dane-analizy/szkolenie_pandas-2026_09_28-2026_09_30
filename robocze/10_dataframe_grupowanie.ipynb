```python
df_sprzedaz = pd.DataFrame(
    {
        "Miesiac": ["Marzec", "Marzec", "Kwiecień", "Kwiecień"],
        "Region": ["Północ", "Południe", "Północ", "Południe"],
        "Produkt": ["Laptop", "Laptop", "Mysz", "Laptop"],
        "Sprzedaz": [4000, 4200, 150, 4100],
    }
)
df_sprzedaz
```

```python
df_sprzedaz.pivot_table(
    index=['Region'],
    columns=['Produkt'],
    values=['Sprzedaz'],
    aggfunc='sum',
)
```

```python
df_sprzedaz.pivot_table(
    index=["Region"],
    columns=["Produkt"],
    values=["Sprzedaz"],
    aggfunc="sum",
    fill_value=0,
)
```

```python
tabela_p = df_sprzedaz.pivot_table(
    index=["Region"],
    columns=["Produkt"],
    values=["Sprzedaz"],
    aggfunc="sum",
    fill_value=0,
    margins=True,
    margins_name='SUMA'
)
tabela_p
```

```python
tabela_p.columns
```

```python
tabela_p.index
```

```python
tabela_p2 = tabela_p.reset_index(drop=False)
tabela_p2
```

```python
movies = pd.read_csv(
    "../dane_przykladowe/kino/static_data.csv",
    sep=";",
    usecols=["movie_title",
             "boxoffice_viewers_weekend_1",
             "boxoffice_viewers_weekend_2",
             "boxoffice_viewers_weekend_3",
             "boxoffice_viewers_weekend_4",
             "boxoffice_viewers_weekend_5",
             "boxoffice_viewers_weekend_6",],
)
movies
```

```python
# grupowanie
df_kadry = pd.DataFrame(
    {
        "Dzial": ["IT", "IT", "IT", "Marketing", "Marketing", "Finanse"],
        "Pracownik": ["Jan", "Anna", "Tomasz", "Ewa", "Piotr", "Zofia"],
        "Pensja": [16000, 12000, 8000, 11000, 7500, 18000],
        "Staz": [5, 3, 1, 4, 2, 8],
    }
)
df_kadry
```

```python
df_kadry.groupby("Dzial").aggregate({"Pensja": ['sum', 'mean'], "Staz": ['min', 'max']})
```

```python
df_kadry.groupby("Dzial", as_index=False).aggregate({"Pensja": ["sum", "mean"], "Staz": ["min", "max"]})
```

```python
df_kadry.groupby("Dzial", as_index=False).aggregate(
    Pensja_suma=('Pensja', 'sum'),
    Srednia_Pensja=('Pensja', 'mean'),
    Staz_min=("Staz", 'min'),
    Staz_max=("Staz", 'max')
)
```
