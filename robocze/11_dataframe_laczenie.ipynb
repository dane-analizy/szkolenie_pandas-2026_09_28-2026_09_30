```python
klienci = pd.DataFrame(
    {"id": [1, 2, 3, 4],
     "nazwisko": ["Kowalski", "Nowak", "Wiśniewski", "Zielińska"]}
)
zamowienia = pd.DataFrame({
    "id_zam": [101, 102, 103],
    "id_klienta": [1, 2, 1,],
    "kwota": [250.0, 490.0, 333.0]})
```

```python
pd.merge(klienci, zamowienia, how='left', left_on='id', right_on='id_klienta')
```

```python
pd.merge(klienci, zamowienia, how="right", left_on="id", right_on="id_klienta")
```

```python
pd.merge(zamowienia, klienci, how="left", left_on="id_klienta", right_on="id")
```

```python
pd.merge(zamowienia, klienci, how="right", left_on="id_klienta", right_on="id")
```

```python
pd.merge(klienci, zamowienia, how="left", left_on="id", right_on="id_klienta", validate="1:1")
```

```python
pd.merge(
    klienci, zamowienia, how="left", left_on="id", right_on="id_klienta", validate="1:m"
)
```

```python
pd.merge(
    klienci,
    zamowienia,
    how="left",
    left_on="id",
    right_on="id_klienta",
    validate="1:m",
    indicator=True,
)
```

```python
pd.merge(zamowienia, klienci, how="right", left_on="id_klienta", right_on="id", indicator=True, )
```

```python
df1 = pd.DataFrame(
    {
        "id": [1, 1, 2, 2],
        "kategoria": ["A", "B", "A", "B"],
        "wartosc_1": [10, 20, 30, 40],
    }
)

df2 = pd.DataFrame(
    {
        "id": [1, 2, 2, 3],
        "kategoria": ["A", "A", "B", "A"],
        "wartosc_2": [100, 200, 300, 400],
    }
)

# Łączenie (domyślnie left inner join) po dwóch kolumnach
wynik = pd.merge(df1, df2, on=["id", "kategoria"])
wynik
```

```python
pd.merge(df1, df2, on="id")
```

```python
wynik2 = pd.merge(df1, df2, left_on=["id", "kategoria"], right_on=["id", "kategoria"])
wynik2
```

```python
# merge_asof - łączenie do najbliższego momentu w czasie
transakcje = pd.DataFrame(
    {
        "czas": pd.to_datetime(["10:00", "10:05", "10:07"]),
        "symbol": ["AAA", "AAA", "BBB"],
        "ilosc": [10, 20, 15],
    }
).sort_values("czas")
transakcje
```

```python
notowania = pd.DataFrame(
    {
        "czas": pd.to_datetime(["09:59", "10:04", "10:06", "10:08"]),
        "symbol": ["AAA", "AAA", "BBB", "BBB"],
        "cena": [150.0, 151.5, 300.0, 301.0],
    }
).sort_values("czas")
notowania
```

```python
pd.merge_asof(transakcje, notowania, on="czas", by="symbol")
```

```python
pd.merge_asof(transakcje, notowania, on="czas", by="symbol", direction='forward')
```

```python
pd.merge_asof(transakcje, notowania, on="czas", by="symbol", direction="nearest")
```

```python
# po staremu, na piechotę
pd.merge(transakcje, notowania, on="symbol") \
    .rename(columns={
        "czas_x": "czas_transakcji",
        "czas_y": "czas_notowania"
    }) \
    .assign(
        delta_czas_s=lambda r: (r["czas_transakcji"] - r["czas_notowania"]).dt.seconds
    ) \
    .sort_values(['symbol', 'ilosc', 'delta_czas_s']) \
    .groupby(['symbol', 'ilosc'], as_index=False) \
    .first()
```
