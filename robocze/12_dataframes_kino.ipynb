```python
import pandas as pd
```

```python
movies = pd.read_csv(
    "../dane_przykladowe/kino/static_data.csv",
    sep=";",
    usecols=["movie_title", "boxoffice_viewers_weekend_1", "countries", "genre_list"],
).dropna()
movies
```

```python
countries_dict = pd.read_csv(
    "../dane_przykladowe/kino/countries.csv",
    sep=";"
)
genres_dict = pd.read_csv("../dane_przykladowe/kino/genres.csv", sep=";")
```

```python
pd.merge(movies, genres_dict, left_on='genre_list', right_on='id')
```

```python
movies['lista_gatunkow'] = movies["genre_list"].str.split(",")
movies = movies.explode("lista_gatunkow")
movies
```

```python
movies = pd.merge(movies, genres_dict, left_on="lista_gatunkow", right_on="id")
movies
```

```python
movies.groupby("name").aggregate(
    Srednia_widownia_na_premierze=("boxoffice_viewers_weekend_1", "mean"),
    Liczba_filmow=("movie_title", "nunique"),
).sort_values("Srednia_widownia_na_premierze", ascending=False)
```

```python
movies[movies["name"] == "Sci-Fi"].sort_values("boxoffice_viewers_weekend_1", ascending=False)
```
