```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.DataFrame(
    {
        "Miesiąc": ["Sty", "Lut", "Mar", "Kwi", "Maj", "Cze"],
        "Zysk": [15, 22, 18, 30, 28, 40],
        "Koszty": [10, 12, 15, 14, 18, 20],
        "Klienci": [100, 150, 130, 200, 190, 250],
    }
)
df
```

```python
plt.figure(figsize=(6, 4))
plt.bar(df['Miesiąc'], df['Zysk'])
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.plot(df["Miesiąc"], df["Zysk"], marker="o")
plt.show()
```

```python
# przygotownie "płótna" na wykres
plt.figure(figsize=(6, 4))

# typ wyrkesy
plt.scatter(df["Zysk"], df["Koszty"], marker="o")

# nazwy osi
plt.xlabel("zysk")
plt.ylabel("koszt")

# skala na osiach
plt.xlim((0, max(df["Zysk"])*1.1))
plt.ylim((0, 25))

# skala logarytmiczna
# plt.xscale("log")

# tytuł wykresu
plt.title("Nasz wykres")

# zapisanie wykresu do pliku
plt.savefig("wykres_2.png", dpi=90)

# pokazanie wykresu
plt.show()

```

```python
# przygotownie "płótna" na wykres
plt.figure(figsize=(6, 4))

plt.bar(df["Miesiąc"], df["Zysk"], color='gray')
plt.scatter(df["Miesiąc"], df["Koszty"], color="#3609d6")

plt.show()
```

```python
# przygotownie "płótna" na wykres
plt.figure(figsize=(6, 4))

# typ wyrkesy
# plt.scatter(df["Zysk"], df["Koszty"], marker="o")
# sns.scatterplot(df, x="Zysk", y="Koszty", size="Klienci", hue='Miesiąc')

sns.barplot(df, x="Miesiąc", y="Koszty", hue="Klienci", palette="GnBu")

# nazwy osi
plt.xlabel("zysk")
plt.ylabel("koszt")

# tytuł wykresu
plt.title("Nasz wykres")


# pokazanie wykresu
plt.show()
```

```python
# przygotownie "płótna" na wykres
plt.figure(figsize=(6, 4))

sns.regplot(df, x="Klienci", y="Koszty")

# pokazanie wykresu
plt.show()
```

```python
import numpy as np
```

```python
dane_klientow = pd.DataFrame(
    {
        "Wiek": np.random.randint(18, 70, 1000),
        "Wydatki_Roczne": np.random.normal(15000, 1000, 1000),
        "Liczba_Transakcji": np.random.randint(2, 40, 1000),
        "Segment": np.random.choice(["VIP", "Lojalny", "Standard"], 1000),
    }
)
dane_klientow
```

```python
plt.figure(figsize=(6, 4))

sns.displot(
    dane_klientow,
    x="Wydatki_Roczne",
)

# pokazanie wykresu
plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.displot(
    dane_klientow,
    x="Wydatki_Roczne", kind='hist', bins=100
)

# pokazanie wykresu
plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.displot(dane_klientow, x="Wydatki_Roczne", kind="hist", binwidth=1000)

# pokazanie wykresu
plt.show()
```

```python
# przygotownie "płótna" na wykres
plt.figure(figsize=(6, 4))

sns.regplot(
    dane_klientow, x="Wiek", y="Wydatki_Roczne", color="lightgray", line_kws={"color": "blue"}
)

# pokazanie wykresu
plt.show()
```

```python
# plotly
fig = px.scatter(dane_klientow, x="Wiek", y="Liczba_Transakcji")

fig.show()
```

```python
fig.write_html("wykres_plotly.html")
```

```python
fig = px.bar(dane_klientow, x="Wiek", y="Liczba_Transakcji", color="Wydatki_Roczne")
fig.show()
```
